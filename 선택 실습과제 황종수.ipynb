### **<선택 실습과제>**
#### 교과서 등에 있는 문제를 마크다운으로 구현하고, 정답을 input으로 받은 뒤 채점 결과를 출력하는 코드를 작성해보자(클라썸 제출).

### **ex)**
#### 다음 그림의 입체도형은 정팔면체이다. 면 ABC와 만나지 않는 면은 모두 몇 개인지 구하여라.

![](https://www.ebsmath.co.kr/innovativeqtb/web_qtb/xip//item2878.jpg)
라.

```python
ans = input("정답 : ")
ans = int(ans)

if ans==1:
    print("정답입니다!")
else:
    print("오답입니다!")
    hint = input("힌트를 제공받으시겠습니까? [예/아니오]")
    if hint=="예":
        print("면 ABC와 마주보는 면을 찾아보세요")
    else:
        print("천천히 다시 풀어보세요")
```
